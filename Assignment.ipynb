```python
import cv2
import numpy as np
```

```python
img=cv2.imread(r"Machine_Learning\OpenCV\Practice.jpg",1)
img=cv2.resize(img,(1200,710))
cv2.namedWindow("Control Panel")
cv2.namedWindow("Output")

def nothing(x):
    pass
  
cv2.createTrackbar("Red","Control Panel",0,255,nothing)
cv2.createTrackbar("Green","Control Panel",0,255,nothing)
cv2.createTrackbar("Blue","Control Panel",0,255,nothing)
cv2.createTrackbar("Brightness","Control Panel",0,100,nothing)
cv2.createTrackbar("Blur","Control Panel",1,100,nothing)
cv2.setTrackbarMin("Blur","Control Panel",1)
cv2.createTrackbar("Contrast","Control Panel",0,127,nothing)
cv2.setTrackbarMin("Contrast","Control Panel",1)
while True:
    r = cv2.getTrackbarPos('Red','Control Panel')
    g = cv2.getTrackbarPos('Green','Control Panel')
    b = cv2.getTrackbarPos('Blue','Control Panel')
    brightness = cv2.getTrackbarPos("Brightness","Control Panel")
    blur=cv2.getTrackbarPos("Blur","Control Panel")
    cont=cv2.getTrackbarPos("Contrast","Control Panel")
    img_1=img.copy()
    #x==b,y==g,z==r
    x,y,z=cv2.split(img_1)
    x=cv2.add(x,b-128)
    y=cv2.add(y,g-128)
    z=cv2.add(z,r-128)  
    bright=(brightness/10)
    brightness_kernel=np.array([
        [0,0,0],
        [0,bright,0],
        [0,0,0]
    ])
    blur_kernel=np.ones((blur,blur))/(blur*blur)
    new_img=cv2.merge([x,y,z])
    new_img=cv2.filter2D(new_img,-1,brightness_kernel)
    new_img=cv2.filter2D(new_img,-1,blur_kernel)
    new_img=cv2.convertScaleAbs(new_img, alpha=cont)
    cv2.imshow("Output",new_img)
    
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
cv2.destroyAllWindows()
```

```python
import cv2

# Read the image. OpenCV reads images in BGR format by default.
img = cv2.imread("Machine_Learning\OpenCV\Practice.jpg",1)

# Split the BGR image into its channels
b, g, r = cv2.split(img)
print((img.T).shape)
# To visualize the channels, you can display each one
cv2.imshow("Blue Channel", b)
cv2.imshow("Green Channel", g)
cv2.imshow("Red Channel", r)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
def bright(x,v):
    f=v/100
    sharpen_kernel=np.array([
        [0,0,0],
        [0,f,0],
        [0,0,0]
    ])
    sharpen_img=cv2.filter2D(x,-1,sharpen_kernel)
    return sharpen_img

img = cv2.imread("Machine_Learning\OpenCV\Practice.jpg",1)
value=150
x=bright(img,value)
cv2.imshow("Identity Image",x)
cv2.imshow("Original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
