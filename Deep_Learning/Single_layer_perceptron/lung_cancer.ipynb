```python
X=lung_cancer_df.drop(["LUNG_CANCER"],axis=1)
Y=lung_cancer_df["LUNG_CANCER"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test= train_test_split(X, Y, test_size= 0.30, random_state=0)
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
lr_model=LogisticRegression(random_state=0)
lr_model.fit(x_train, y_train)
```

```python
y_lr_pred= lr_model.predict(x_test)
y_lr_pred
```

```python
from sklearn.metrics import classification_report
lr_cr=classification_report(y_test, y_lr_pred)
print(lr_cr)
```

### Single Layer Perceptron

```python
import tensorflow as tf
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
```

```python
model=tf.keras.Sequential([tf.keras.layers.Dense(1,activation="sigmoid",input_shape=(15,))])
"""
tf.keras.Sequential([tf.keras.layers.Dense(number_of_neurons,activation_function,input_shape)])
number_of_neurons = number of neurons in the layer
activation_function = what activation function to be used
input_shape = number of columns/features the model should except/shape of input feature
(number,)(8,)= each input has n number of freatures
(number,number) (28,28)= each input has (n*n) features 
(number,number)(100,1) = each input is sequence of 100 timesteps of 1 feature each
"""
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
"""
model.compile(optimizer,loss_function,metrics)
optimizer = which optimization technique to be used 
loss = which loss function to be used 
metrics = model evaluation metrics
"""
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
hist=model.fit(x_train,y_train,epochs=300,batch_size=30,validation_split=0.3,verbose=0)
```

```python
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Accuracy: {accuracy:.4f}")
print("Loss -> ",loss)
```
