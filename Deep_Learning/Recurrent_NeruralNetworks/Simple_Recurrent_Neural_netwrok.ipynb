```python
chars=sorted(list(set(input_text)))     # Creating a list of unique words in ascending order
char_to_index={char:i for i ,char in enumerate(chars)} # -> dict   # maps each word with unique index based on position in sorted list 
index_to_char={i: char for i ,char in enumerate(chars) } # -> dict # creates reverse dicionary 
```

```python
seq_length=5
sequences=[]
labels=[]
 
for i in range(len(input_text)-seq_length):                     # Extarcts every possible 5 length sequence 
    seq=input_text[i:i+seq_length]                              # Extracts 5 chars from i
    label=input_text[i+seq_length]                              # Extracts the char that comes just after the sequence
    sequences.append([char_to_index[char] for char in seq])     # converts the chars in it's index value and store them in list of indices
    labels.append(char_to_index[label])                         # convert label char into it's index and store it
    
X=np.array(sequences)   
Y=np.array(labels)
```

```python
X_onehot=tf.one_hot(X,len(chars))
Y_onehot=tf.one_hot(Y,len(chars))
```

```python
model=Sequential()
model.add(SimpleRNN(100,input_shape=(seq_length,len(chars)),activation="relu",return_sequences=True))
"""
model.add(SimpleRnn(number_of_nodes,input_shape,activation,return_sequences))
arg1 -> number of RNN blocks in the layer
arg2 -> input shape (only for first layer)
arg3 -> activation function 
arg4 -> (flag) when we stack multiple layers ,bydefault False
        True -> returns full sequences at each timestamp
        False -> returns only the final output """
model.add(SimpleRNN(100,return_sequences=True,activation='tanh'))
model.add(SimpleRNN(50,activation="relu"))
model.add(Dense(len(chars),activation="softmax"))
```

```python
model.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
print(model.summary())
```

```python
model.fit(X_onehot,Y_onehot,epochs=100)
```

```python
start_seq= "wormholes"
generated_text=start_seq

for i in range(200):
    x=np.array([[char_to_index[char] for char in generated_text[-seq_length:]]])        # predictiong next word using the last 5 words
    x_onehot=tf.one_hot(x,len(chars))       # converts integer vectors into one hot vectors 
    prediction=model.predict(x_onehot)      
    next_index=np.argmax(prediction)        # choose the char with highest probability
    next_char=index_to_char[next_index]     # converts the predicted index into actual chars
    generated_text+=next_char               # appends the char into output str

print("Generated Text:")
print(generated_text)
```

### Using PyTorch

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
```

```python
seq_length=5
sequences=[]
labels=[]
 
for i in range(len(input_text)-seq_length):                     # Extarcts every possible 5 length sequence 
    seq=input_text[i:i+seq_length]                              # Extracts 5 chars from i
    label=input_text[i+seq_length]                              # Extracts the char that comes just after the sequence
    sequences.append([char_to_index[char] for char in seq])     # converts the chars in it's index value and store them in list of indices
    labels.append(char_to_index[label])                         # convert label char into it's index and store it
    
X = torch.tensor(sequences, dtype=torch.long)      # (num_samples, seq_len)
Y = torch.tensor(labels, dtype=torch.long)

num_samples = X.size(0)
seq_len = X.size(1)
num_chars = len(chars)

# one-hot → (num_samples, seq_len, num_chars)
X_onehot = torch.nn.functional.one_hot(X, num_classes=num_chars).float()

# RNN expects → (seq_len, batch, input_size)
X_onehot = X_onehot.permute(1, 0, 2)
```

```python
class SimpleRNN(nn.Module):
    def __init__(self, ):
        super().__init__()
        self.rnn1=nn.RNN(input_size=num_chars,hidden_size=50,num_layers=15,nonlinearity="relu")
        """
        nn.RNN(input_size,hidden_size,num_layers,nonlinearity,bias,batch_first,dropout,bidirectional
        input_size -> number of expected features of input
        hidden_size -> number of features of hidden layer
        num_layers -> number of RNN layers
        nonlinearity -> activation function
        bias(flag) -> if "False' bias is not added, by default "True"
        batch_first(flag) -> by default "False"
                            if "True" output tensors = (batch, seq, feature) 
                            if "False" output tensors = (seq, batch, feature).
        dropout -> by default 0
                    if "Non Zero" adds an dropout layer
                    if "0" doesn't add dropout layer
        bidirectional – If "True" bidirectional RNN. by default "False"
        """
        self.fc=nn.Linear(50,len(chars))
        """
        nn.Linear(in_features,out_features,bias)
        in_features –> size of each input sample
        out_features –> size of each output sample
        bias (flag) –> If set to False, the layer will not learn an additive bias. Default: True
        """

    def forward(self,x):
        h0=torch.zeros(15,x.size(1),50)
        """
        torch.zeros(number of time steps,batch_size,feature dimension) -> creates an tensor with passed dimensions
        """
        out,_=self.rnn1(x,h0)
        out=self.fc(out[-1])
        return out 

model=SimpleRNN()
```

```python
print(model)
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(),lr=0.001)
```

```python
num_epoch=100
for epoch in range(num_epoch):
    model.train()
    epoch_loss=0
    outputs=model(X_onehot)
    loss=criterion(outputs,Y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    epoch_loss+=loss.item()

    print(f"Epoch[{epoch+1}/{num_epoch}], Loss: [{epoch_loss/len(chars)}]")
```

```python
# Create index_to_char mapping (reverse of char_to_index)
index_to_char = {idx: char for char, idx in char_to_index.items()}

start_seq = "porta"  # Must be exactly seq_length (5) characters
generated_text = start_seq

model.eval()  # Set model to evaluation mode

with torch.no_grad():  # Disable gradient calculation for inference
    for i in range(200):
        # Get last seq_length characters
        x = [char_to_index[char] for char in generated_text[-seq_length:]]
        x = torch.tensor([x], dtype=torch.long)  # Shape: (1, seq_length)
        
        # Convert to one-hot encoding
        x_onehot = torch.nn.functional.one_hot(x, num_classes=num_chars).float()
        # Shape: (1, seq_length, num_chars)
        
        # Permute to (seq_length, batch, num_chars) for RNN
        x_onehot = x_onehot.permute(1, 0, 2)
        
        # Get prediction
        prediction = model(x_onehot)  # Shape: (1, num_chars)
        
        # Get the character with highest probability
        next_index = torch.argmax(prediction, dim=1).item()
        next_char = index_to_char[next_index]
        
        # Append to generated text
        generated_text += next_char

print("Generated Text:")
print(generated_text)
```
